```python
import pyspark

from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("Filter Practice").getOrCreate()
```

```python
df_pyspark = spark.read.csv("Members.csv", header=True, inferSchema=True)

df_pyspark.show()
```

```python
from pyspark.sql import functions as F

df_salary50000 = df_pyspark.select(["firstName","lastName","Age","Salary"]).where(F.col("Salary") >= 50000)
df_salary50000.show()
```

```python
df_salary50000v2 = df_pyspark.filter("Salary <= 50000")
df_salary50000v2.show()
```

```python
df_salaryv3 = df_pyspark.filter((df_pyspark['Salary'] == 50000) | (df_pyspark['Salary'] == 60000))
df_salaryv3.show()
```

```python
df_salarNot50000 = df_pyspark.filter(~(df_pyspark["Salary"] == 50000))
df_salarNot50000.show()
```
