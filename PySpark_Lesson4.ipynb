**PySpark GroupBy and Aggregate Functions**

```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("PySpark Group Function").getOrCreate()
```

```python
df_employee = spark.read.csv("employee_list.csv", header = True, inferSchema = True)
df_employee.show()
```

```python
df_employee.printSchema()
```

```python
df_avesalary = df_employee.groupBy("Department").sum()
df_avesalary.show()
```

```python
df_maxsalary = df_employee.groupBy("Department").count()
df_maxsalary.show()
```

```python
df_sumSalary = df_employee.agg({"Salary": "sum"})
df_sumSalary.show()
```

```python
df_meanSalary = df_employee.agg({"Salary": "mean"})
df_meanSalary.show()
```

```python
df_maxsalary1 = df_employee.groupBy("Department").max().orderBy("max(Salary)", ascending = False)
df_maxsalary1.show()
```
