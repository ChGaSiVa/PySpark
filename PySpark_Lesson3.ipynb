```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("PySparkLesson3").getOrCreate()
```

```python
df_pyspark = spark.read.csv("Members.csv", header=True, inferSchema=True)
```

```python
df_pyspark.show()
```

```python
df_new = df_pyspark.drop('middleName')
df_new.show()
```

```python
df_filledna = df_pyspark.na.fill({'middleName': 'N/A'})
df_filledna.show()
```

```python
df_dropedna = df_pyspark.dropna()
df_dropedna.show()
```

```python
df_droppednullMiddleName = df_pyspark.dropna(subset=['middleName'])
df_droppednullMiddleName.show()
```

```python
df_pyspark.show()
```

```python
df_cleaned = df_pyspark.dropna(how='any', thresh=6)
df_cleaned.show()
```

```python
df_cleaned2 = df_pyspark.dropna( subset = ["Age", "Department"], thresh = 1)
df_cleaned2.show()
```

```python
df_pyspark.show()
```

```python
df_fillednaAge = df_pyspark.fillna(20, subset=['Age'])
df_fillednaAge.show()
```

```python
from pyspark.ml.feature import Imputer

imputer = Imputer(
    inputCols=["Age", "Salary"],
    outputCols=["{}_imputed".format(c) for c in ["Age", "Salary"]]
).setStrategy("mean")

df_imputed = imputer.fit(df_pyspark).transform(df_pyspark)

df_imputed.show()
```
