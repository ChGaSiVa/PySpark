```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("Refresher Test").getOrCreate()
```

```python
df_pyspark = spark.read.csv("PySpark_Refresher.csv", header=True, inferSchema=True)
df_pyspark.show()
```

```python
df_first3 = df_pyspark.select("firstName", "lastName", "Salary")
df_first3.show(3)
```

```python
from pyspark.sql import functions as F

df_newColumn = df_pyspark.withColumn("FullName", F.concat_ws(" ", *[F.col(c) for c in ["firstName", "lastName"]]))
df_newColumn.show(5)
```

```python
df_dropna = df_pyspark.dropna(how = "any")
df_dropna.show()
```

```python
df_dropnaAll = df_pyspark.dropna(how = "all")
df_dropnaAll.show()
```

```python
df_dropnaThresh = df_pyspark.dropna(thresh=6)
df_dropnaThresh.show()
```

```python
print(df_dropnaThresh.count())
```

```python
df_filteredRows = df_pyspark.where((F.col("Department") == "Sales") & (F.col("Experience") > 3))
df_filteredRows.show() 
```

```python
df_dropSpecific = df_pyspark.dropna(subset=["middleName","Salary"], how = "any")
df_dropSpecific.show()
```

```python
df_fillNA = df_pyspark.fillna({"middleName": "N/A", "Salary": 50000, "Age": 30})
df_fillNA.show()
```
