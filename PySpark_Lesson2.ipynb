```python
import pyspark

from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("PySparkLesson2").getOrCreate()
```

```python
df_pyspark = spark.read.csv("People.csv", header = True, inferSchema= True)
df_pyspark.show(5)
```

```python
df_pyspark.printSchema()
```

```python
df_pyspark.select("*").show()
```

```python
df_filtered = df_pyspark.select("firstName", "middleName", "lastName")

df_filtered.limit(3).show()
```

```python
df_filtered.dtypes
```

```python
df_filtered.printSchema()
```

```python
df_new = df_pyspark.withColumn('ID Adjusted', df_pyspark['ID'] + 2)

df_new.show(10)
```

```python
df_new = df_new.drop('id')

df_new.show(10)
```

```python
df_updated = df_new.withColumnRenamed('ID Adjusted', 'id_updated')

df_updated.show(10)
```
