```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.master("local[1]").appName("PracticeSession").getOrCreate()
```

```python
df_emp = spark.read.csv("employees.csv", header=True, inferSchema=True)
df_emp.show()
```

```python
from pyspark.sql import functions as F

df_empNewColumn = df_emp.withColumn("FullName", F.concat_ws(" ", *[F.col(c) for c in ("firstName", "lastName")]))

df_empNewColumn.show()
```

```python
df_filteredEmployees = df_emp.where((F.col("Department") == "IT") & (F.col("Salary") >= 50000))
df_filteredEmployees.show()
```
